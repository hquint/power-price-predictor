```python
import os
from dotenv import load_dotenv
from entsoe import EntsoePandasClient
import pandas as pd
from datetime import datetime, timedelta

# Load .env file
load_dotenv()

# Get API key from environment
api_key = os.getenv("ENTSOE_API_KEY")
client = EntsoePandasClient(api_key=api_key)
```

```python
country_code = 'DE_LU'  # Germany

# Define time range
start = pd.Timestamp(datetime.now() - timedelta(days=7), tz='Europe/Berlin')
end = pd.Timestamp(datetime.now(), tz='Europe/Berlin')
```

```python
# Query actual total load
df_load = client.query_load(country_code, start=start, end=end)

# Display
df_load.plot(title="Germany - Actual Load (MW)", figsize=(12, 4))
df_load.head()
```

### day ahead prices

```python
df_price = client.query_day_ahead_prices(country_code, start=start, end=end)

df_price.plot(title="Germany - Day-Ahead Prices (€/MWh)", figsize=(12, 4))
```

### load forecast

```python
df_forecast = client.query_load_forecast(country_code, start=start, end=end)

df_forecast.plot(title="Germany - Load Forecast", figsize=(12, 4))
```
